```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.ticker as ticker
from scipy.stats import mannwhitneyu
import matplotlib.patches as mpl_patches
```

```python
df=pd.read_table('Fig4e.txt')
df.head()
```

```python
from matplotlib.ticker import MaxNLocator
import matplotlib.ticker as ticker
boxprops = {'edgecolor': 'k', 'linewidth': 1,}
lineprops = {'color': 'k', 'linewidth': 1}
boxplot_kwargs = dict({'boxprops': boxprops, 'medianprops': lineprops,
                       'whiskerprops': lineprops, 'capprops': lineprops,
                       'width': 0.75,})
swarmplot_kwargs = dict({'linewidth': 0.6, 'size': 5})
Order=['Col-0','Col-0-mgr5',
       'S1-1','S1-1-mgr5',
       'S15-3','S15-3-mgr5']
cm=1/2.54
fig, axes = plt.subplots(figsize=(8*cm,5*cm))
sns.boxplot(x=df['Sample'], 
                y='Mg',
                color='grey',
                data=df,
                order=Order,
                linewidth=1.5,
                **boxplot_kwargs)
sns.swarmplot(x=df['Sample'],
              y='Mg', 
              data=df,
              order=Order,
              color="0",
              s=3,
              **swarmplot_kwargs)
axes.set_xlabel('')
axes.set_xticklabels([])
plt.ylabel('')
plt.xlabel('')

sns.despine(trim= True)
plt.tight_layout()
# plt.savefig("Fig4e.pdf", format="pdf")
```

### Mann-Whitney-U tests

```python
mannwhitneyu(df.loc[df['Sample'] == 'Col-0', 'Mg'],
            df.loc[df['Sample'] == 'Col-0-mgr5', 'Mg'])
```
