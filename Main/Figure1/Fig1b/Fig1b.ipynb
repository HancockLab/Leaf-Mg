```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
from scipy.stats import spearmanr
```

```python
df = pd.read_csv('810_Mg_blues_top3SNPs_wCoordinates_variables.csv')
df
```

```python
df=df.dropna(subset=['Latitude'])
```

```python
from matplotlib.ticker import MaxNLocator
import matplotlib.ticker as ticker
cm=1/2.54
# fig,axes = plt.subplots(figsize=(8*cm,5*cm))
fig = plt.figure(figsize=(8*cm,6*cm))
ax = sns.regplot(data=df,
            x='Latitude',
            y='Mg_blues',
            color='grey',
            scatter_kws={'color': 'grey','clip_on':False}, 
            line_kws={'color': 'r'})
r,p = spearmanr(df['Mg_blues'],df['Latitude'])
plt.text(15, 5800, '$\itR$ = {:.2f}\n$P$ = {:.2e}\n$n$ = {}'.format(r,p,df.shape[0]), fontsize=12)
# plt.ylim(3000,7000)
# plt.xlim(3200,7000)
plt.ylabel('Leaf [Mg] (µg/g DW)',fontsize=10)
# plt.xlabel()
ax.xaxis.set_major_locator(ticker.MultipleLocator(5))
# ax.yaxis.set_major_locator(ticker.MultipleLocator(15))
sns.despine(offset=10, trim=True)
plt.savefig("Fig1b.pdf", format="pdf",bbox_inches="tight",facecolor='w')
```

```python
from matplotlib.ticker import MaxNLocator
import matplotlib.ticker as ticker
cm=1/2.54
# fig,axes = plt.subplots(figsize=(8*cm,5*cm))
fig = plt.figure(figsize=(9*cm,8*cm))
ax = sns.scatterplot(data=df,
            x='Latitude',
            y='Mg_blues',
            color='grey',
            edgecolor='none',
            marker='o',
            s=50,
            alpha=0.7)
r,p = spearmanr(df['Mg_blues'],df['Latitude'])
plt.text(10, 5800, '$\itR$ = {:.2f}\n$P$ = {:.2e}\n$n$ = {}'.format(r,p,df.shape[0]), fontsize=12)
# plt.ylim(3000,7000)
# plt.xlim(3200,7000)
plt.ylabel('Leaf [Mg] (µg/g DW)',fontsize=10)
ax.set_xlim(10,70)
# plt.xlabel()
ax.xaxis.set_major_locator(ticker.MultipleLocator(10))
# ax.yaxis.set_major_locator(ticker.MultipleLocator(15))
sns.despine(offset=10, trim=True)
plt.savefig("Fig1b.pdf", format="pdf",bbox_inches="tight",facecolor='w')
```
