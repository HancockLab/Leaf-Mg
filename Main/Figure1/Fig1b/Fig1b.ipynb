```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
from scipy.stats import spearmanr
```

```python
df=pd.read_table('Europe_677_accession.txt')
df.head()
```

```python
from matplotlib.ticker import MaxNLocator
import matplotlib.ticker as ticker
cm=1/2.54
# fig,axes = plt.subplots(figsize=(8*cm,5*cm))
fig = plt.figure(figsize=(8*cm,6*cm))
ax = sns.regplot(data=df,
            x='Latitude',
            y='Mg25',
            color='grey',
            scatter_kws={'color': 'grey','clip_on':False}, 
            line_kws={'color': 'r'})
r,p = spearmanr(df['Mg25'],df['Latitude'])
plt.text(38, 5800, '$\itR$ = {:.2f}\n$P$ = {:.2e}\n$n$ = {}'.format(r,p,df.shape[0]), fontsize=12)
# plt.ylim(3000,7000)
# plt.xlim(3200,7000)
plt.ylabel('Leaf [Mg] (µg/g DW)',fontsize=10)
# plt.xlabel()
ax.xaxis.set_major_locator(ticker.MultipleLocator(5))
# ax.yaxis.set_major_locator(ticker.MultipleLocator(15))
sns.despine(offset=10, trim=True)
# plt.savefig("Fig1b.pdf", format="pdf",bbox_inches="tight",facecolor='w')
```
