```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
df = pd.read_csv('810_Mg_blues_top3SNPs_wCoordinates_variables.csv')
df
```

```python
df.columns.to_list()
```

```python
df=df.rename(columns={'coarse_fragments':'coarse frag',
                      'bulk_density':'bulk density',
                      'CaCO3':'CaCO$_3$',
                      'ai':'AI',
                      'awc':'AWC',
                      'GRIDCODE':'Mg(Soil)',
                      'Mg_blues':'Mg(Leaf)',
                      'srad_mean':'srad',
                     })
```

```python
df=df.drop(columns=['chr5_21392702',
                    'chr5_21405867',
                    'chr5_282011',
                    'CLASSLIMIT',
                    'Accession_ID'])
df
```

```python
df=df.dropna()
df
```

```python
df=df.loc[(df['Latitude']>36.52) & 
             (df['Latitude']<68.80) & 
             (df['Longitude']>-8.54) & 
             (df['Longitude']<38.28)]
df
```

```python
fig = plt.figure(figsize=(20,20))
mask = np.triu(np.ones_like(df.corr(), dtype=bool))
cmap = sns.diverging_palette(230, 20, as_cmap=True)
sns.heatmap(df.corr(method='spearman'), 
            mask=mask, 
            vmax=1, 
            vmin=-1,
            center=0, 
            cmap='bwr',
            square=True, 
            annot=True,
            fmt=".1f",
            linewidths=.2, 
            cbar_kws={"shrink": .5})
plt.savefig("FigS1.pdf", format="pdf",facecolor="w",bbox_inches="tight")
```
